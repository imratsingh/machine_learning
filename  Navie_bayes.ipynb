Naive Bayes = A probabilistic classifier based on Bayes’ Theorem with an assumption that features are independent
it is used for classification problems and it used divide data into the different differnt categories

1. Gaussian Naive Bayes
Used when features are continuous values
Assumes data follows a normal (Gaussian) distribution

👉 Example:
Height, weight, marks, temperature

2. Multinomial Naive Bayes
Used for discrete data
Mostly used in text classification

👉 Example:
Spam detection
Word frequency in documents

3. Bernoulli Naive Bayes

Used for binary data (0 or 1)

Focuses on presence/absence of features

👉 Example:

Whether a word appears in a document (yes/no)
