# polinomial Regression

Polynomial Regression is a type of regression where the relationship
between input (X) and output (Y) is not a straight line, but a curve.

y = B0 + B1X1 + B2X2pawer2 + B3X3pawer3.....BnXnpawern

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\polynomial_regression.csv")
dataset.head(3)
```

```python
dataset.drop("Position",axis=1,inplace=True)
```

```python
dataset.head(10)
```

```python
dataset.isnull().sum()
```

```python
plt.scatter(dataset["Level"],dataset["Salary"])
plt.xlabel("level")
plt.ylabel("salary")
plt.show()
```

```python
dataset.corr()
```

```python
x = dataset[["Level"]]
y = dataset["Salary"]
```

we wont to convert data polynomial from 

degree takes as your vish

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
pf = PolynomialFeatures(degree=2)
pf.fit(x)
```

```python
x = pf.transform(x)
```

then we wont split data into train and test split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)*100
```

```python
#equation ==>  y = M1X1+M2X2^2+M3+X3^3+c
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
prd = lr.predict(x)
```

```python
plt.scatter(dataset["Level"],dataset["Salary"])
plt.plot(dataset["Level"],prd,c="red")
plt.xlabel("Level")
plt.ylabel("Salary")
plt.legend(["ord","prd"])
plt.show()
```
