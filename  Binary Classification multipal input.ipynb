# Binary Classification multipal input
Binary Classification (Multiple Input)
Multiple features (2 या ज्यादा)
Example:
CGPA + Score → Placement (0/1)

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "cgpa": [5, 5.5, 6, 6.5, 8, 8.5, 9, 9.5],
    "score": [50, 55, 60, 65, 80, 85, 90, 95],
    "placement": [0, 0, 0, 0, 1, 1, 1, 1]
}

dataset = pd.DataFrame(data)
print(dataset)
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x="cgpa",y="score",data = dataset,hue="placement")
plt.legend(loc=1)
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["placement"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)*100
```

```python
lr.predict([[8.5,80]])
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf=lr)
```
