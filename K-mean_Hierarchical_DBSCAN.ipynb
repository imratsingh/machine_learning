
# Unsupervised learning 
is a type of machine learning in which the model is 
trained on unlabeled data and it learns to find hidden patterns, structures,
or relationships in the data without any prior guidance or output labels.

1. K-Means Clustering

Definition:
K-Means is an unsupervised algorithm that divides data into K clusters based on similarity.

work with = linearly seprable data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\Iris_logistic regression_multiclass_classification.csv")
dataset.head(3)
```

```python
dataset.head(3)
dataset.shape
# dataset["SepalLengthCm"].shape
```

```python
sns.pairplot(data=dataset)
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
for i in range(2,21):
    kn = KMeans(n_clusters=i,init='k-means++')
    kn.fit(dataset)
    wcss.append(kn.inertia_)
```

```python
plt.figure(figsize=(10,5))
plt.plot([i for i in range(2,21)],wcss,marker="o")
plt.xticks([i for i in range(2,21)])
plt.xlabel("no of cluster")
plt.ylabel("wcss")
plt.grid(axis="x")
plt.show()
```

```python
kmn = KMeans(n_clusters=3)
dataset["predict"] = kmn.fit_predict(dataset)
```

```python
sns.pairplot(data=dataset,hue="predict")
plt.savefig("predict")
plt.show()
```

# Hierarchical clustering

Hierarchical clustering is an unsupervised learning algorithm that groups data into clusters by creating a tree-like structure called a dendrogram.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\Iris_logistic regression_multiclass_classification.csv")
dataset.head()
```

```python
dataset.drop(columns=["Id","Species"],inplace=True)
```

```python
# sns.pairplot(data=dataset)
# plt.show()
```

```python
import scipy.cluster.hierarchy as sc
```

```python
plt.figure(figsize=(10,20))

sc.dendrogram(sc.linkage(dataset,method="single",metric="euclidean"))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
ac = AgglomerativeClustering(n_clusters=2,linkage="single")
dataset["predict"] = ac.fit_predict(dataset)
```

```python
dataset.head(2)
```

```python
sns.pairplot(data=dataset,hue="predict")
plt.show()
```

# DBSCAM clustering

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons
```

```python
x,y = make_moons(n_samples=250,noise=0.05)
```

```python
df = {"data1":x[:,0],"data2":x[:,1]}
```

```python
dataset = pd.DataFrame(df)
```

```python
dataset.head(3)
```

```python
# scatterplot used for check data linear or non-linear
sns.scatterplot(x="data1",y="data2",data=dataset)
plt.show()
```

```python
from sklearn.cluster import DBSCAN
```

```python
db = DBSCAN(eps=0.4,min_samples=5)
dataset["predict"] = db.fit_predict(dataset)
```

```python
db.fit_predict(dataset)
```

```python
sns.scatterplot(x="data1",y="data2",data=dataset,hue="predict")
plt.show()
```
