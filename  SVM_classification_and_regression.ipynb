# SVM(support vector machine) for classification
data linear seprebal

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\logistic_regression_student_placement.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
dataset.shape
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x="cgpa",y="iq",data=dataset,hue="placement")
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["placement"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=40)
```

```python
from sklearn.svm import SVC
```

```python
{'linear', 'poly', 'rbf', 'sigmoid', 'precomputed'}
```

```python
sv = SVC(kernel="linear")
sv.fit(x_train,y_train)
```

```python
sv.score(x_test,y_test)*100,sv.score(x_train,y_train)*100
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf=sv)
plt.show()
```

# SVM(support vector machines) for Regression (data linear form)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\placement.csv")
dataset.head(3)
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x="cgpa",y="package",data=dataset)
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["package"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.svm import SVR
```

```python
sv = SVR(kernel="linear")
sv.fit(x_train,y_train)
```

```python
sv.score(x_test,y_test)*100,sv.score(x_train,y_train)*100
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
sns.scatterplot(x="cgpa",y="package",data=dataset)
plt.plot(dataset["cgpa"],sv.predict(x),color="red")
plt.show()
```
