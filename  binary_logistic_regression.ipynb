```python
What is Logistic Regression?

Logistic Regression is a supervised machine learning algorithm used for classification problems.

    there are three types of logistic regression 
(1).binary logistic regression
(2).multinomial logistic regression
(3).ordinal logistic regression

. Binary Logistic Regression

👉 Used when there are only 2 classes

Examples:

Yes / No
0 / 1
Spam / Not Spam
2. Multinomial Logistic Regression

👉 Used when there are more than 2 classes

Examples:

Cat / Dog / Cow
Red / Blue / Green

👉 No specific order in classes

3. Ordinal Logistic Regression

👉 Used when classes have a natural order

Examples:

Low / Medium / High
Poor / Average / Good

👉 Order matters here
```

#  Binary Logistic Regression

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\Social_Network_Ads_logistic_regression.csv")
dataset.head(3)
```

```python
dataset.shape
```

```python
dataset.drop(columns=["User ID","Gender","EstimatedSalary"],inplace=True)
```

```python
dataset.head(3)
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x="Age",y="Purchased",data = dataset)
plt.show()
```

```python
x = dataset[["Age"]]
y = dataset["Purchased"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)*100
```

```python
lr.predict([[40]])
```

```python
plt.figure(figsize=(5,4))
sns.scatterplot(x = "Age",y="Purchased",data = dataset)
sns.lineplot(x="Age",y=lr.predict(x),data = dataset,color = "red")
plt.show()
```
