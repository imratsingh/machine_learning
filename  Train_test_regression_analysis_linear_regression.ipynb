```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\placement.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x = "cgpa",y = "package",data = dataset)
plt.show()
```

```python
x = dataset[["cgpa"]]
y = dataset["package"]
```

```python
x.ndim
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
y = 0.57425647 * x -1.0270069374542108
```

```python
lr.predict([[6.89]])
```

```python
lr.score(x_test,y_test)*100
```

```python
 0.57425647 * 6.89 -1.0270069374542108
```

```python
y_prd = lr.predict(x)
```

```python
plt.figure(figsize=(5,4))
sns.scatterplot(x = "cgpa",y="package",data = dataset,label="org")
plt.plot(dataset["cgpa"],y_prd,c="red",label="predictionline")
plt.legend(["org","predictionline"])
plt.savefig("predict.jpg")
plt.show()
```
