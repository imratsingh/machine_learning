# multilinear Regrassion

-- Multiple Linear Regression (often called multilinear regression) is an extension of simple linear regression where we use more than one independent variable to predict a single output.
    
Multiple Linear Regression predicts the value of y using two or more input variables (x₁, x₂, x₃ …)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\multi_linear_regression.csv")
dataset.head(3)
```

```python
dataset.rename(columns={"income":"Salary"},inplace=True)
```

```python
dataset.isnull().sum()
```

```python
sns.pairplot(data = dataset)
plt.show()
```

```python
sns.heatmap(data = dataset.corr(),annot=True)
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["Salary"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)*100
```

```python
lr.coef_
```

```python
lr.intercept_
```
