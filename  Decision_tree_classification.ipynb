 # Decision tree 

```python
decision tree is a supercised machine techinique that can be used for both classification and regression problem but mostly it is preferred for solving classification problems

in order to build a tree we use the CART algorithms which stands for classification and regression tree algorithm
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\Social_Network_Ads_logistic_regression.csv")
dataset.head(3)
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
# non-
sns.scatterplot(x="Age",y="EstimatedSalary",data=dataset,hue="Purchased")
plt.show()
```

```python
dataset.isnull().sum()
```

```python
dataset.drop(columns=["User ID","Gender"],inplace=True)
```

```python
dataset.head(3)
```

```python
x = dataset.iloc[:,:-1]
y = dataset["Purchased"]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
sc.fit(x)
x = pd.DataFrame(sc.transform(x),columns=x.columns)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
```

# pre-pruning max_depth=5

```python
dt = DecisionTreeClassifier(max_depth=5)
dt.fit(x_train,y_train)
```

```python
dt.score(x_test,y_test)*100,dt.score(x_train,y_train)*100
```

```python
dt1 = DecisionTreeClassifier(criterion="entropy")
dt1.fit(x_train,y_train)
```

```python
dt1.score(x_test,y_test)*100
```

```python
dt.score(x_test,y_test)*100,dt.score(x_train,y_train)*100
```

```python
dt.predict([[35,20000]])
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf=dt)
plt.show()
```

```python
from sklearn.tree import plot_tree
```

```python
plt.figure(figsize=(25,25))
plot_tree(dt)
plt.savefig("demo.jpg")
plt.show()
```

# post-pruning

```python
for i in range(1,20):
    dt2=DecisionTreeClassifier(max_depth=i)
    dt2.fit(x_train,y_train)
    print(dt2.score(x_test,y_test),dt2.score(x_train,y_train))
```

```python
dt = DecisionTreeClassifier(max_depth=3)
dt.fit(x_train,y_train)
```

```python
dt.score(x_test,y_test)*100 ,dt.score(x_train,y_train)*100
```

```python
plt.figure(figsize=(25,25))
plot_tree(dt)
plt.savefig("demo.jpg")
plt.show()
```
