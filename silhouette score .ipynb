```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv(r"C:\Users\imrat\OneDrive\Desktop\data_science_all_uses_items\Data_set\Iris_logistic regression_multiclass_classification.csv")
dataset.head(3)
```

```python
dataset.drop(columns=["Id","Species"],inplace=True)
```

```python
sns.pairplot(data=dataset)
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
for i in range(2,21):
    kn = KMeans(n_clusters=i,init='k-means++')
    kn.fit(dataset)
    wcss.append(kn.inertia_)
```

```python
plt.figure(figsize=(10,5))
plt.plot([i for i in range(2,21)],wcss,marker="o")
plt.xticks([i for i in range(2,21)])
plt.xlabel("no of cluster")
plt.ylabel("wcss")
plt.grid(axis="x")
plt.show()
```

```python
ss = []
no_c = [j for j in range(2,21)]
for i in range(2,21):
    km1 = KMeans(n_clusters=i)
    km1.fit(dataset)
    ss.append(silhouette_score(dataset,km1.labels_))
```

```python
plt.plot(no_c,ss)
plt.xlabel("no of cluster")
plt.ylabel("silhouette core")
plt.xticks(no_c)
plt.grid(axis="x")
plt.show()
```
